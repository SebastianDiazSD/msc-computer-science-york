## Exercise 1 & 2: Regex patterns and test program

All five patterns are implemented and tested against `The_Raven.txt` in the cells below.

````python
import re

# Load the file
with open('The_Raven.txt', 'r', encoding='utf-8') as f:
    text = f.read()

print("=== Source text ===")
print(text)
````

### Task 1: Find the word 'shrieked'

**Pattern:** `\bshrieked\b`

- `\b` = word boundary — ensures we match the whole word, not a substring (e.g. would not match 'unshrieked' if it existed)
- `shrieked` = literal characters

````python
pattern_shrieked = r'\bshrieked\b'
matches = re.findall(pattern_shrieked, text)

print(f"Pattern: {pattern_shrieked}")
print(f"Matches found: {matches}")
print(f"Count: {len(matches)}")

# Show line context
for line in text.splitlines():
    if re.search(pattern_shrieked, line):
        print(f"  -> Found in line: {line.strip()}")
````

### Task 2: Find the word 'bleak'

**Pattern:** `\bbleak\b`

This is a word-boundary search identical to Task 1. The word 'bleak' does not appear in this excerpt — this is the expected and correct result. The regex itself is valid; the text simply does not contain the word.

````python
pattern_bleak = r'\bbleak\b'
matches = re.findall(pattern_bleak, text)

print(f"Pattern: {pattern_bleak}")
print(f"Matches found: {matches}")
print(f"Count: {len(matches)}")
print("Note: 'bleak' does not appear in this excerpt of The Raven.")
````

### Task 3: Count words containing 'pp'

**Pattern:** `\b\w*pp\w*\b`

- `\b` = word boundary (start and end)
- `\w*` = zero or more word characters before and after 'pp'
- `pp` = literal double-p substring

This matches any complete word that contains 'pp' anywhere within it.

````python
pattern_pp = r'\b\w*pp\w*\b'
matches = re.findall(pattern_pp, text)

print(f"Pattern: {pattern_pp}")
print(f"Matches: {matches}")
print(f"Count: {len(matches)}")

# Unique words (without duplicates)
print(f"Unique words with 'pp': {sorted(set(matches))}")
````

### Task 4: Replace all `!` with `#`

**Pattern:** `!`  (literal exclamation mark)

Uses `re.sub()` to perform the substitution. The original text is not modified — the result is stored in `text_replaced`.

````python
pattern_exclamation = r'!'
text_replaced = re.sub(pattern_exclamation, '#', text)

# Count how many replacements were made
original_count = len(re.findall(pattern_exclamation, text))
replaced_count = text_replaced.count('#')

print(f"Pattern: {pattern_exclamation!r}")
print(f"Exclamation marks in original: {original_count}")
print(f"Hash symbols in result: {replaced_count}")
print()
print("=== Modified text ===")
print(text_replaced)
````

### Task 5: Words starting with 't' that do NOT end with 'e' (case-insensitive)

**Pattern:** `\b[tT]\w*[^eE\W]\b`  
**Equivalent with flag:** `\b[t]\w*[^e]\b` with `re.IGNORECASE`

Breaking it down:
- `\b` = word boundary
- `[tT]` = starts with lowercase or uppercase 't'
- `\w*` = zero or more word characters in the middle
- `[^eE\W]` = last character is NOT 'e', NOT 'E', and NOT a non-word character
- `\b` = word boundary

**Edge case:** single-letter word 't' — the pattern requires at least 2 characters because `[tT]` consumes the first and `[^eE\W]` must match a different second character. Words like 'the' (ends in 'e') are correctly excluded.

````python
# Using re.IGNORECASE flag — cleaner and more readable
pattern_t_not_e = r'\bt\w*[^e]\b'
matches = re.findall(pattern_t_not_e, text, re.IGNORECASE)

print(f"Pattern: {pattern_t_not_e!r} with IGNORECASE flag")
print(f"All matches: {matches}")
print(f"Unique matches: {sorted(set(w.lower() for w in matches))}")
print(f"Count (including duplicates): {len(matches)}")
print()

# Verify exclusions — these should NOT appear:
print("Verification — words starting with t that END with e (should be excluded):")
should_be_excluded = re.findall(r'\bt\w*e\b', text, re.IGNORECASE)
print(f"  {sorted(set(w.lower() for w in should_be_excluded))}")
````

## Exercise 3: Discussion

### Advantages and disadvantages of regular expressions

**Advantages:**
- **Concise** — a single pattern replaces many lines of character-by-character logic
- **Powerful** — handles complex matching rules (boundaries, negations, groups, repetition) that are difficult to code manually
- **Universal** — the same regex syntax works across Python, JavaScript, Java, grep, sed, and most modern languages
- **Built into standard libraries** — no extra dependencies (`import re` in Python)
- **Fast** — the regex engine is compiled C code under the hood, faster than equivalent pure Python string loops

**Disadvantages:**
- **Readability** — complex patterns (e.g. email validation) are nearly unreadable without comments or decomposition
- **Debugging difficulty** — a single wrong character (e.g. `\B` instead of `\b`) silently changes behaviour
- **Not suitable for structured formats** — never use regex to parse HTML, XML, or JSON; use proper parsers (`xml.etree`, `json`) which handle edge cases correctly
- **Greedy by default** — `.*` matches as much as possible, often more than intended; requires `.*?` for non-greedy matching
- **No context** — regex matches patterns, not meaning. It cannot distinguish a date like `12/03/2026` from a version number `12/03/2026`

### Where regex would have been beneficial in my own experience

In my **App 2 — Tender Document Risk Analyzer**, I use PyMuPDF to extract raw text from German construction tender PDFs. German contract documents (VOB/B) contain clause references like `§ 4 Abs. 7`, `§ 8 Nr. 3`, `§ 13 Abs. 1`. 

Without regex, finding these required splitting on whitespace and checking each token — fragile and verbose. With regex:
```python
pattern = r'§\s*\d+\s*(?:Abs\.?|Nr\.?)\s*\d+'
clauses = re.findall(pattern, contract_text)
```
This finds all clause references in one line, regardless of spacing variations introduced by PDF extraction. Regex was the right tool precisely because clause references follow a consistent syntactic pattern, even when the surrounding text varies.
