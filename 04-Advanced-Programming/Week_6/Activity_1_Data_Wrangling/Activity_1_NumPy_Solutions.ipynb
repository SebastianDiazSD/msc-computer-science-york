## Exercise 1 – Random Array, Square Root, Sum

```python
# Generate 10,000 random numbers in a 2D array (100x100)
random_array = np.random.rand(100, 100)
print("Array shape:", random_array.shape)
print("First 3 values (flattened):", random_array.flatten()[:3])

# Calculate the square root of each value
sqrt_array = np.sqrt(random_array)

# Sum all the square root values
total = np.sum(sqrt_array)
print(f"\nSum of all square roots: {total:.4f}")

# One-liner equivalent:
total_oneliner = np.sum(np.sqrt(np.random.rand(100, 100)))
print(f"One-liner total: {total_oneliner:.4f}")
```

## Exercise 3 – Sales Data Analysis with NumPy

File: `Activity_1_Exercise_3_SalesData.csv`  
Stores: L3, L1, P2, N6, N4, B8  
Months: Apr-18 to Mar-19 (12 months)

```python
import csv

# Load the CSV, separating row headers, column headers, and data
with open('Activity_1_Exercise_3_SalesData.csv', 'r') as f:
    reader = csv.reader(f)
    rows = list(reader)

# Extract headers
col_headers = np.array(rows[0][1:])   # Month labels
row_headers = np.array([row[0] for row in rows[1:]])  # Store labels
data_values = np.array([[int(v) for v in row[1:]] for row in rows[1:]], dtype=float)

print("Column headers (months):", col_headers)
print("Row headers (stores):", row_headers)
print("\nData array shape:", data_values.shape)
print(data_values)
```

```python
# --- Second Quarter: Jul-18 to Sep-18 (inclusive) → column indices 3, 4, 5 ---
q2_indices = [3, 4, 5]   # Jul-18, Aug-18, Sep-18
q2_months = col_headers[q2_indices]
q2_data = data_values[:, q2_indices]

print("Second Quarter (Jul-18 to Sep-18):\n")
print(f"{'Month':<10} {'Total':>8} {'Mean':>8}")
print("-" * 30)
for i, month in enumerate(q2_months):
    col = q2_data[:, i]
    print(f"{month:<10} {np.sum(col):>8.0f} {np.mean(col):>8.2f}")
```

```python
# --- L3 and L1: First Quarter (Apr-18 to Jun-18) and Third Quarter (Oct-18 to Dec-18) ---
# Store row indices: L3=0, L1=1
l3_idx = np.where(row_headers == 'L3')[0][0]
l1_idx = np.where(row_headers == 'L1')[0][0]

q1_cols = slice(0, 3)   # Apr-18, May-18, Jun-18
q3_cols = slice(6, 9)   # Oct-18, Nov-18, Dec-18

for store_idx, store_name in [(l3_idx, 'L3'), (l1_idx, 'L1')]:
    for q_cols, q_name in [(q1_cols, 'Q1 (Apr-18 to Jun-18)'), (q3_cols, 'Q3 (Oct-18 to Dec-18)')]:
        q_data = data_values[store_idx, q_cols]
        print(f"{store_name} – {q_name}")
        print(f"  Values : {q_data}")
        print(f"  Total  : {np.sum(q_data):.0f}")
        print(f"  Mean   : {np.mean(q_data):.2f}\n")
```

```python
# --- N6 and N4: Sales lower than 400, with months listed ---
n6_idx = np.where(row_headers == 'N6')[0][0]
n4_idx = np.where(row_headers == 'N4')[0][0]

for store_idx, store_name in [(n6_idx, 'N6'), (n4_idx, 'N4')]:
    store_data = data_values[store_idx, :]
    mask = store_data < 400
    low_months = col_headers[mask]
    low_values = store_data[mask]
    print(f"{store_name} – Sales below 400:")
    for month, val in zip(low_months, low_values):
        print(f"  {month}: {val:.0f}")
    print()
```

## Exercise 4 – Random Walk Problem

A bug lives in the middle of a 7-unit island. If it reaches either end it drowns. Movement is random (left or right). Test over 100 iterations and apply statistics.

```python
def random_walk_bug(island_length=7, iterations=100):
    """
    Simulate a random walk on a 1D island.
    - Island positions: 0 to island_length (edges = drowning)
    - Bug starts at the centre
    - Returns an array of step counts for each iteration
    """
    step_counts = np.zeros(iterations, dtype=int)
    start_pos = island_length // 2  # centre = position 3

    for i in range(iterations):
        position = start_pos
        steps = 0

        while 0 < position < island_length:
            # Randomly choose -1 (left) or +1 (right)
            move = np.random.choice([-1, 1])
            position += move
            steps += 1

        step_counts[i] = steps

    return step_counts

np.random.seed(42)   # for reproducibility
step_counts = random_walk_bug(island_length=7, iterations=100)

print("Random Walk – Bug on 7-unit Island (100 iterations)")
print(f"  Min steps   : {np.min(step_counts)}")
print(f"  Max steps   : {np.max(step_counts)}")
print(f"  Mean steps  : {np.mean(step_counts):.2f}")
print(f"  Median steps: {np.median(step_counts):.1f}")
print(f"  Std dev     : {np.std(step_counts):.2f}")
print(f"  25th pctile : {np.percentile(step_counts, 25):.1f}")
print(f"  75th pctile : {np.percentile(step_counts, 75):.1f}")
print(f"\nFirst 20 step counts: {step_counts[:20]}")
```

## Exercise 5 – numpy.sort() Algorithms: Summary

`numpy.sort()` supports three sorting algorithms via the `kind` parameter:

| Algorithm | `kind=` | Time (avg) | Stable? | Memory |
|---|---|---|---|---|
| Quicksort | `'quicksort'` (default) | O(n log n) | No | O(log n) |
| Mergesort | `'mergesort'` | O(n log n) | Yes | O(n) |
| Heapsort | `'heapsort'` | O(n log n) | No | O(1) |
| Timsort | `'stable'` | O(n log n) | Yes | O(n) |

**Recommendations:**

- **Quicksort** – Best general-purpose choice for **large, unsorted numerical datasets**. Fastest in practice due to cache efficiency, though worst-case is O(n²). Suitable for most data science use cases where stability is not required.

- **Mergesort** – Use when a **stable sort is required** (preserves relative order of equal-valued elements). Important when sorting records by a secondary key after a primary sort. Higher memory usage (O(n)) is a trade-off.

- **Heapsort** – Use when **memory is tightly constrained** and a worst-case O(n log n) guarantee is needed. Slower in practice than Quicksort due to poor cache performance. Rarely the first choice.

- **Timsort (`'stable'`)** – NumPy's alias for a stable sort, backed by mergesort. Best choice for **real-world data that is partially sorted**, as it exploits existing order. Preferred over `'mergesort'` when working with DataFrames or structured arrays.

**Conclusion**: For most datasets in data analysis, `'quicksort'` (the default) is the most practical. Switch to `'stable'` when sort order matters for equal elements.
